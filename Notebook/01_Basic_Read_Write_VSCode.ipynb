# 01 — Basic Read & Write Operations

**Run in VS Code.** Keep `shopping_data.csv` in the same folder as this notebook.

This notebook demonstrates basic Python file handling: reading, writing, appending, and copying files. It does not require Pandas or NumPy.

```python
from pathlib import Path

# The CSV should be in the same folder as this notebook.
DATA_FILE = Path("shopping_data.csv")

if not DATA_FILE.exists():
    raise FileNotFoundError(
        "shopping_data.csv was not found. Put it in the same folder as this notebook."
    )

print("Found:", DATA_FILE.resolve())
```

## 1. Write a text file

```python
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("Shopping Data Project\\n")
    file.write("This file demonstrates basic write operations.\\n")

print("sample.txt created.")
```

## 2. Read a text file

```python
with open("sample.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

## 3. Append new content

```python
with open("sample.txt", "a", encoding="utf-8") as file:
    file.write("This line was appended later.\\n")

with open("sample.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 4. Read the first five lines of the CSV

```python
with open(DATA_FILE, "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(line.rstrip())
        if line_number == 5:
            break
```

## 5. Create a byte-for-byte copy of the CSV

```python
copy_file = Path("shopping_data_copy.csv")

with open(DATA_FILE, "rb") as source, open(copy_file, "wb") as destination:
    destination.write(source.read())

print("Copy created:", copy_file.resolve())
```

## 6. Confirm that the files exist

```python
for path in [DATA_FILE, Path("sample.txt"), copy_file]:
    print(f"{path.name}: exists={path.exists()}, size={path.stat().st_size} bytes")
```
