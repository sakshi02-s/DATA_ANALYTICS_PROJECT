# 02 — Data Analysis Using Pandas & NumPy

**Run in VS Code.** Keep `shopping_data.csv` in the same folder as this notebook.

This notebook loads and explores the shopping dataset, checks data quality, calculates numerical summaries, and demonstrates filtering and grouping.

```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_FILE = Path("shopping_data.csv")
if not DATA_FILE.exists():
    raise FileNotFoundError(
        "shopping_data.csv was not found. Put it in the same folder as this notebook."
    )

df = pd.read_csv(DATA_FILE)
print("Dataset loaded successfully.")
```

## 1. Preview and understand the dataset

```python
display(df.head())
print("Rows and columns:", df.shape)
print("\\nColumn names:")
print(df.columns.tolist())
print("\\nData types:")
print(df.dtypes)
print("\\nDataset information:")
df.info()
```

## 2. Generate descriptive statistics

```python
display(df.describe(include="all").T)
```

## 3. Check missing values and duplicates

```python
print("Missing values by column:")
display(df.isna().sum().to_frame("Missing_Count"))

print("Exact duplicate rows:", df.duplicated().sum())
print("Repeated Order_Id rows:", df["Order_Id"].duplicated(keep=False).sum())
display(df[df["Order_Id"].duplicated(keep=False)].sort_values("Order_Id"))
```

## 4. Use NumPy for numerical summaries

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    values = df[column].dropna().to_numpy()
    if values.size:
        print(f"\\n{column}")
        print("  Mean:", np.mean(values))
        print("  Median:", np.median(values))
        print("  Minimum:", np.min(values))
        print("  Maximum:", np.max(values))
```

## 5. Explore category counts

```python
for column in ["City", "Product_Category", "Payment_Mode", "Customer_Type"]:
    print(f"\\n{column}")
    display(df[column].value_counts(dropna=False).to_frame("Count"))
```

## 6. Filter and group the data

```python
average_order_value = df["Order_Value_Rs"].mean()
above_average = df[df["Order_Value_Rs"] > average_order_value]

print("Average order value:", round(average_order_value, 2))
print("Orders above average:", len(above_average))
display(above_average.head())

city_summary = (
    df.groupby("City", dropna=False)["Order_Value_Rs"]
      .mean()
      .sort_values(ascending=False)
      .to_frame("Average_Order_Value_Rs")
)
display(city_summary)
```

## 7. Save an analysis copy

```python
analysis_copy = Path("shopping_data_analysis_copy.csv")
df.to_csv(analysis_copy, index=False)
print("Analysis copy saved to:", analysis_copy.resolve())
```
